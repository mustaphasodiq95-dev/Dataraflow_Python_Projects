```python
import json

Students = [
    {"name": "Favour", "age": "43", "grade": "A"},
    {"name": "Adeyemi", "age": "23", "grade": "C"}
]


with open("Students.txt", "w") as f:
    f.write(json.dumps(Students))


with open("Students.txt", "r") as f:
    data = f.read()
    Students_from_file = json.loads(data)

print(Students_from_file)
```

```python
import random

numbers = range(1, 11)
operators = ["+", "-", "*", "/"]

score = 0

for question in range(5):
    x = random.choice(numbers)
    operator = random.choice(operators)
    y = random.choice(numbers)

    if operator == "+":
        correct_answer = x + y
    elif operator == "-":
        correct_answer = x - y
    elif operator == "*":
        correct_answer = x * y
    elif operator == "/":
        correct_answer = x / y

    user_answer = float(input(f"What is {x} {operator} {y}? "))

    if user_answer == correct_answer:
        print("Correct!")
        score += 1
    else:
        print(f"Wrong! The correct answer is {correct_answer}")

print(f"You scored {score}/5")
```

```python
import datetime

try:
    birthday_input = input("Enter your birthday (DD-MM): ")

# have to give python a year so it accomodate every possible date, i used 2000 because its a leap year might as well use 2008,2012 or 2016
    birthday = datetime.datetime.strptime(
        birthday_input + "-2000",
        "%d-%m-%Y"
    )

    today = datetime.datetime.now()

    birthday_this_year = birthday.replace(year=today.year)

    if birthday_this_year < today:
        birthday_next = birthday_this_year.replace(year=today.year + 1)
    else:
        birthday_next = birthday_this_year

    days_left = (birthday_next - today).days

    print(f"There are {days_left} days until your next birthday.")

except ValueError:
    print("Invalid date. Please enter your birthday in DD-MM format, e.g. 04-09.")
```
