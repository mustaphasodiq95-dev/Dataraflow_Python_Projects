```python
my_cart = {}
item = ""

while item != "done":
    item = input()

    if item == "done":
        break

    price = int(input())
    my_cart[item] = price
total_cost = sum(my_cart.values())

print(total_cost)
```

```python
import random

number = random.randint(1, 20)

while True:
    guess = int(input("Guess the number: "))
    if guess > number:
        print("Too high")
    elif guess < number:
        print("Too low")
    else:
        print("Correct!")
        break
```

```python
students = []

while True:
    name = input("Enter student name (or done to finish): ")

    if name == "done":
        break

    students.append(name)

with open("students.txt", "w") as file:
    for student in students:
        file.write(student + "\n")

with open("students.txt", "r") as file:
    content = file.read()

print(content)
```

6. A local variable is defined inside a function and can generally only be accessed within that function, while a global variable is defined outside functions and can be accessed from different parts of the program.

Example of local variable
def greet():
    name = "Sodiq"
    print(name)

name = "Sodiq" is a local variable because it belongs to greet().

Example of global variable
name = "Sodiq"

def greet():
    print(name)

name is global because it was created outside the function, and the function can access it.

```python
numbers = [1, 2, 3, 4, 5, 6]

def even_numbers(numbers):
    result = []

    for number in numbers:
        if number % 2 == 0:
            result.append(number)

    return result

print(even_numbers(numbers))
```
